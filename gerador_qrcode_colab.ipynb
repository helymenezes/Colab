```python
import qrcode
from IPython.display import display

# Link para gerar o QR Code
link = input(" Insira o link para gerar o QR Code: ")
nome_fig = input("informe o nome do arquivo para salvar: ")

# Criar QR Code
qr = qrcode.QRCode()
qr.add_data(link)
qr.make(fit=True)

# Mostrar QR Code no Jupyter Notebook
img = qr.make_image(fill_color="black", back_color="white")
display(img)

# Salvar a imagem do QR Code
img.save(f"/content/drive/MyDrive/Colab Notebooks/qrcode_projetos_eletricos/{nome_fig}.png")  # Salva como "qr_code.png" no diretório atual
```
