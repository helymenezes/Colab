```python
import pytube

# Insira a URL do vídeo do YouTube
url =  'https://youtu.be/Qng8zhFmPA4?si=_Nxp3JYcWbt7yfLZ'

# Crie um objeto YouTube Downloader
yt = pytube.YouTube(url)

# Obtenha o título do vídeo
titulo = yt.title

# Extraia apenas o áudio do vídeo
audio = yt.streams.filter(only_audio=True).first()

# Defina o caminho de download
caminho_download = f"{titulo}.mp3"

# Baixe o áudio do vídeo
audio.download(filename=caminho_download)

print(f"Música '{titulo}' baixada com sucesso!")
```

```python
import yt_dlp

# Insira a URL do vídeo do YouTube
url = 'https://youtu.be/YuoW2sLdwJ4?si=TssmX4V-HJSN3IeW'

# Define as opções de download
ydl_opts = {
    'format': 'bestvideo[ext=mp4]+bestaudio[ext=m4a]/mp4',
    'outtmpl': '%(title)s.%(ext)s',
    'http_headers': {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3'
    }
}

# Baixa o vídeo
with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    ydl.download([url])

print("Vídeo baixado com sucesso!")
```
